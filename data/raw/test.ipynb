```python

# Các video có tags bị thiếu (NaN)
train_df=pd.read_csv("youtube_train.csv")
missing_tags = train_df.loc[
    train_df["tags"].isna() & train_df["tag_count"]>0,
    ["video_id"]
]

print(missing_tags.count())
```

```python
test_df=pd.read_csv("youtube_test.csv")
missing_tags = test_df.loc[
    test_df["tags"].isna() & test_df["tag_count"]>0,
    ["video_id"]
]

print(missing_tags.count())
```
